```python
schema_str = "id INT, customer_name STRING, product STRING, quantity INT, price DOUBLE, remarks STRING"

df = spark.read.schema(schema_str).option("header","true").csv("/Volumes/new_practice/my_files/csv_files/sales_no_header.csv")

display(df)
```

```python
df = spark.read.option("header","true").option("inferSchema","true").option("delimiter","|").csv("/Volumes/new_practice/my_files/csv_files/sales_pipe.csv")
df.show(truncate=False)
```

```python
df = spark.read.option("header","true").option("inferSchema","true").option("delimiter","|").option("quote","'").csv("/Volumes/new_practice/my_files/csv_files/sales_pipe_single_quote.csv")

df.show(truncate=False)
```

```python
df = spark.read.option("header","true").option("inferSchema","true").option("delimiter","|").option("quote","'").option("escape","#").csv("/Volumes/new_practice/my_files/csv_files/sales_pipe_single_quote_escape.csv")

df.show(truncate=False)
```
