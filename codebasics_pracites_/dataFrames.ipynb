##Column Filtering

```python
df_trimmed = df.select("title","imdb_rating","industry")
df_trimmed.show(3,truncate=True)
```

##Row Filtering

```python
df_filtered = df.filter((df.release_year >= 2000) & (df.release_year <= 2010))
df_filtered.show(20, truncate=False)
```

```python
from pyspark.sql.functions import col

df_filtered = df.filter((col("release_year") >= 2000) & (col("release_year") <= 2010))
df_filtered.show(20, truncate=False)
```

```python
df_filtered = df.filter(col("release_year").between(2000,2010))
display(df_filtered)
```

```python
df_marvel = df.filter(col("studio")=="Marvel Studios")
display(df_marvel)
```

```python
df.select("industry").distinct().show(truncate=False)
```

```python
df.select("language").distinct().show(truncate=False)
```

```python
df.withColumn("Profit",col("revenue")-col("budget")).show(truncate=False)
```

```python
df.withColumnRenamed("revenue","Total_revenue").show(truncate=False)
```
